```python
from datasets import load_dataset

sst2 = load_dataset("nyu-mll/glue", "sst2")
print(sst2)
print(sst2["train"][0])
```

```python
from collections import Counter
for split in ["train", "validation", "test"]:
  print(split, Counter(sst2[split]["label"]))
```

```python
imdb = load_dataset("stanfordnlp/imdb")
print(imdb)
print(imdb["train"][0]["text"][:300], "...")
print("label:", imdb["train"][0]["label"])
```

```python
import numpy as np
lens = [len(t.split()) for t in imdb["train"]["text"][:2000]]
print("IMDB words/review: mean", np.mean(lens), "| median", np.median(lens), "| max", max(lens))
```
